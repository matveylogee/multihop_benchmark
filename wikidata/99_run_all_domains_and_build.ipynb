# Run All Domains and Build Dataset

Master orchestrator notebook that executes all 15 domain-specific generation notebooks in sequence and then triggers the final dataset assembly step.

**Produces:**
- Per-domain `.jsonl` files under `out_wikidata_benchmark/domain_outputs/`
- Final merged benchmark dataset via `98_final_dataset_assembly.ipynb`

```python
from pathlib import Path
exec(Path("common_helpers.py").read_text(encoding="utf-8"), globals())
exec(Path("notebook_runner.py").read_text(encoding="utf-8"), globals())

DOMAIN_NOTEBOOKS = [
    "01_cinema.ipynb",
    "02_geo_ru.ipynb",
    "03_geo_international.ipynb",
    "04_books.ipynb",
    "05_videogames.ipynb",
    "06_music_albums.ipynb",
    "07_software.ipynb",
    "08_people.ipynb",
    "09_scientists_mathematicians.ipynb",
    "10_paintings_museums.ipynb",
    "11_spacecraft.ipynb",
    "12_countries.ipynb",
    "13_dishes.ipynb",
    "14_smartphones.ipynb",
    "15_universities_airports_cars.ipynb",
]

for nb_path in DOMAIN_NOTEBOOKS:
    print("\n" + "=" * 100)
    print("Running", nb_path)
    run_ipynb(nb_path, globals())

print("\n" + "=" * 100)
print("Running final assembly")
run_ipynb("98_final_dataset_assembly.ipynb", globals())
```
